## **Messwerte**

```python
import numpy as np
import uncertainties as uc
import uncertainties.unumpy as unp

U_batt = 1.367 #Spannung der Batterie in V

U_0_data = 5 #Ausgangsspannung zum Vergleichen in V
U_0_fehler = 0.5
U_0 = uc.ufloat(U_0_data, U_0_fehler)

R_1_data = np.array([256, 518, 780, 18, 46, 72, 100, 126, 388, 650])  + 10
R_1_fehler = 10
R_1 = unp.uarray(R_1_data, R_1_fehler)

R_2_data = np.array([100, 200, 300, 10, 20, 30, 40, 50, 150, 250])
R_2_fehler = 10
R_2 = unp.uarray(R_2_data, R_2_fehler)
```

## **Berechnung der Batteriespannung**

```python

U_x = U_0 * (R_2 / (R_1 + R_2))

print(f"Berechnete Batteriespannung: U_x = {U_x.mean()} V")
```
