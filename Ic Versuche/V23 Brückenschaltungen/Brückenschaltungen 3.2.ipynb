## **Berechnungen für zwei Widerstände**

```python
import numpy as np
import uncertainties as uc
import uncertainties.unumpy as unp

R_1_data = 675 #Fest gewählter Widerstand in Ohm
R_1_fehler = 10
R_1 = uc.ufloat(R_1_data, R_1_fehler)

"""-----------------------------------------------------------------------------------------------------------------------------------------------------------"""

R_3_data = np.array([100, 200, 300, 400, 500, 600, 700, 800, 900, 1000])
R_3_fehler = 10
R_3 = unp.uarray(R_3_data, R_3_fehler)

R_4_data = np.array([68, 142, 214, 288, 360, 432, 506, 580, 652, 726])
R_4_fehler = 10
R_4 = unp.uarray(R_4_data, R_4_fehler)

"""-----------------------------------------------------------------------------------------------------------------------------------------------------------"""
#Berechnungen

R = ((R_4 * R_1) / R_3).mean()

print(f"Widerstand A: R = {R} Ohm")
```

## **Messwerte für zwei Reihenschaltungen**

```python
C_1_data = 476e-9 #Fest gewählte Kapazität in F
C_1_fehler = 0.01*C_1_data
C_1 = uc.ufloat(C_1_data, C_1_fehler)

R_1_data = 174 #Fest gewählter Widerstand in Ohm
R_1_fehler = 10
R_1 = uc.ufloat(R_1_data, R_1_fehler)

"""-----------------------------------------------------------------------------------------------------------------------------------------------------------"""

R_3_data = np.array([100, 200, 300, 400, 500, 600, 700, 150, 250, 350])
R_3_fehler = 10
R_3 = unp.uarray(R_3_data, R_3_fehler)

R_4_data = np.array([138, 278, 420, 558, 700, 840, 1690, 209, 350, 488])
R_4_fehler = 10
R_4 = unp.uarray(R_4_data, R_4_fehler)

"""-----------------------------------------------------------------------------------------------------------------------------------------------------------"""

C = ((C_1 * R_3) / R_4).mean()
R = ((R_1 * R_4) / R_3).mean()

print(f"Kapazität A: C_A = {C*(10**9)} nF")
print(f"Widerstand A: R_A = {R} Ohm \n")
```

## **Ausschlagverfahren**

```python
import scipy
import matplotlib.pyplot as plt
import matplotlib

matplotlib.use("pgf")
matplotlib.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'arial',
    'text.usetex': True,
    'pgf.rcfonts': False,
})


def f(x, k, c):
    return k*x +c

R_4_data = np.array([312, 322, 332, 342, 352, 362, 372, 382, 392, 402, 412, 422, 432, 442, 452, 462])
R_4_fehler = 10
R_4 = unp.uarray(R_4_data, R_4_fehler)

U_data = np.array([0.220, 175.7e-3, 126.1e-3, 82.4e-3, 36e-3, 0, -0.04, -85.5e-3, -136.6e-3, -171.5e-3, -0.210, -0.247, -0.281, -0.318, -0.363, -0.393])
U_fehler = 0.01
U = unp.uarray(U_data, U_fehler)

coefficients, pcov = scipy.optimize.curve_fit(f, unp.nominal_values(R_4), unp.nominal_values(U), sigma= unp.std_devs(U), absolute_sigma= True)
k = uc.ufloat(coefficients[0], np.sqrt(pcov[0,0]))
c = uc.ufloat(coefficients[1], np.sqrt(pcov[1,1]))

print(f"Steigung der Geraden: k = {k*10**3}\n")
print(f"Y-Achsenabschnitt der Geraden: c = {c}\n")

#Plot
fig, ax0 = plt.subplots(1,1, figsize = (4,4))

ax0.errorbar(unp.nominal_values(R_4), unp.nominal_values(U) , yerr= unp.std_devs(U), fmt="o", label="Messwerte", color="blue", markersize=4, capsize=3, ecolor="r")

xwerte = np.linspace(np.min(R_4_data)*0.99, np.max(R_4_data)*1.01, 1000)
ax0.plot(xwerte, f(xwerte, k.nominal_value, c.nominal_value), color = "black", linestyle = "--", label = "Linearer Fit")

ax0.set_xlabel("Widerstand  $R_4\\,$ $[\\Omega ]$")
ax0.set_ylabel("Spannung $U$ [V]")
ax0.grid(True)
ax0.set_title("Ausschlagmethode")
ax0.legend()
plt.tight_layout()
plt.savefig("Ausschlag.pgf")
```
