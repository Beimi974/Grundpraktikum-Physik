## **Messwerte**

```python
import numpy as np


T = np.array([38, 110, 120, 135, 150, 160, 170, 180, 190, 197, 200, 202]) #Temperatur

U = np.array([45, 41, 40, 34, 32, 29, 26, 23, 18, 4, 2, 1]) #Spannung in mV
U_Fehler = 5
```

## **Plot**

```python
import scipy.optimize
import matplotlib.pyplot as plt

import matplotlib
matplotlib.use("pgf")
matplotlib.rcParams.update({
    "pgf.texsystem": "pdflatex",
    'font.family': 'arial',
    'text.usetex': True,
    'pgf.rcfonts': False,
})

fig , ax1 = plt.subplots(1,1, figsize = (4,4))

ax1.errorbar(T, U, yerr= U_Fehler, fmt="o", label="Messdaten", color="blue", markersize=3, capsize=3, ecolor="r")

ax1.hlines(0, xmin= 0, xmax = 210, color = "black")


def f(x,k,c):
    return k*x + c

coefficients8, pcov8 = scipy.optimize.curve_fit(f, T[10:], U[10:], sigma= U_Fehler, absolute_sigma= True)
kappa_8 = coefficients8[0]
kappa_8_Fehler = np.sqrt(pcov8[0,0])
c_8 = coefficients8[1]
xwerte = np.linspace(180, 210, 500)


ax1.plot(xwerte, kappa_8*xwerte + c_8, color = "blue", linestyle = "--", label = "Linearer Fit nahe der Nullstelle")

ax1.set_xlim(xmin = 0, xmax = 210)
ax1.set_xlabel(r"Temperatur $T$ (°C)")
ax1.set_ylabel(r"Remenanz (mV)")
ax1.grid(True)
ax1.set_title("Messdaten Hysterese")
ax1.legend()

plt.savefig("Curie.pgf")
```
