## **Kopplungsgrad aus linearer Regression**

```python
import scipy.optimize
import matplotlib.pyplot as plt
import sympy as sp
from Skripte.Fehlerfortpflanzung import Gaußfehler

N_Ratio = N_2 / N_1

#Frequenz 1

U1p, U2p = sp.symbols("U_1, U_2")
expr3 = U2p / U1p
display(expr3)
Variablen = np.array([U1p, U2p])

#Wiederstand 1
U_Ratio_f1_R1 = U2_f1_R1 / U1_f1_R1
U_Ratio_f1_R1_Fehler = np.zeros(len(U_Ratio_f1_R1), dtype = float)

for i in range(len(U_Ratio_f1_R1_Fehler)):
    U_Ratio_f1_R1_Fehler[i] = Gaußfehler(expr3, Variablen, np.array([U1_f1_R1[i], U2_f1_R1[i]]), np.array([U1_f1_R1_Fehler, U2_f1_R1_Fehler]))

#Wiederstand 2
U_Ratio_f1_R2 = U2_f1_R2 / U1_f1_R2
U_Ratio_f1_R2_Fehler = np.zeros(len(U_Ratio_f1_R2), dtype = float)

for i in range(len(U_Ratio_f1_R2_Fehler)):
    U_Ratio_f1_R2_Fehler[i] = Gaußfehler(expr3, Variablen, np.array([U1_f1_R2[i], U2_f1_R2[i]]), np.array([U1_f1_R2_Fehler, U2_f1_R2_Fehler]))

#Lineare Regression

def f(x,k,c):
    return k*x +c #Fit Funktion

coefficients1, pcov1 = scipy.optimize.curve_fit(f, N_Ratio, U_Ratio_f1_R1, sigma= U_Ratio_f1_R1_Fehler, absolute_sigma= True)
kappa_1 = coefficients1[0]
kappa_1_Fehler = np.sqrt(pcov1[0,0])
c_1 = 0

print(f"Kopplungsgrad für f_1 = {f_1} und Wiederstand 1: kappa = {kappa_1} +/- {kappa_1_Fehler}")

coefficients2, pcov2 = scipy.optimize.curve_fit(f, N_Ratio, U_Ratio_f1_R2, sigma= U_Ratio_f1_R2_Fehler, absolute_sigma= True)
kappa_2 = coefficients2[0]
kappa_2_Fehler = np.sqrt(pcov2[0,0])
c_2 = 0

print(f"Kopplungsgrad für f_1 = {f_1} und Wiederstand 2: kappa = {kappa_2} +/- {kappa_2_Fehler}")

"""-------------------------------------------------------------------------------------------------------------------------------------------------------------------"""

#Frequenz 2

N_Ratio = N_2 / N_1

U1p, U2p = sp.symbols("U_1, U_2")
expr3 = U2p / U1p
display(expr3)
Variablen = np.array([U1p, U2p])

#Wiederstand 1
U_Ratio_f2_R1 = U2_f2_R1 / U1_f2_R1
U_Ratio_f2_R1_Fehler = np.zeros(len(U_Ratio_f2_R1), dtype = float)

for i in range(len(U_Ratio_f2_R1_Fehler)):
    U_Ratio_f2_R1_Fehler[i] = Gaußfehler(expr3, Variablen, np.array([U1_f2_R1[i], U2_f2_R1[i]]), np.array([U1_f2_R1_Fehler, U2_f2_R1_Fehler]))

#Wiederstand 2
U_Ratio_f2_R2 = U2_f2_R2 / U1_f2_R2
U_Ratio_f2_R2_Fehler = np.zeros(len(U_Ratio_f2_R2), dtype = float)

for i in range(len(U_Ratio_f2_R2_Fehler)):
    U_Ratio_f2_R2_Fehler[i] = Gaußfehler(expr3, Variablen, np.array([U1_f2_R2[i], U2_f2_R2[i]]), np.array([U1_f2_R2_Fehler, U2_f2_R2_Fehler]))

#Lineare Regression

def f(x,k,c):
    return k*x +c #Fit Funktion

coefficients3, pcov3 = scipy.optimize.curve_fit(f, N_Ratio, U_Ratio_f2_R1, sigma= U_Ratio_f2_R1_Fehler, absolute_sigma= True)
kappa_3 = coefficients3[0]
kappa_3_Fehler = np.sqrt(pcov3[0,0])
c_3 = 0

print(f"Kopplungsgrad für f_2 = {f_2} und Wiederstand 1: kappa = {kappa_3} +/- {kappa_3_Fehler}")

coefficients4, pcov4 = scipy.optimize.curve_fit(f, N_Ratio, U_Ratio_f2_R2, sigma= U_Ratio_f2_R2_Fehler, absolute_sigma= True)
kappa_4 = coefficients4[0]
kappa_4_Fehler = np.sqrt(pcov4[0,0])
c_4 = 0

print(f"Kopplungsgrad für f_2 = {f_2} und Wiederstand 2: kappa = {kappa_4} +/- {kappa_4_Fehler}")
```
